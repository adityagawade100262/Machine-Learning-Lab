# 1. NumPy

## Introduction

NumPy (Numerical Python) is the fundamental library used for numerical computations in Python. It provides support for multidimensional arrays, matrices, and mathematical operations. NumPy is faster and more memory-efficient than Python lists because it stores data in contiguous memory. It offers a wide range of mathematical, statistical, and linear algebra functions that simplify complex calculations. Many popular libraries such as Pandas, SciPy, TensorFlow, and Scikit-Learn are built on top of NumPy. Due to its speed, efficiency, and powerful features, NumPy is considered the foundation of Machine Learning, Data Science, Artificial Intelligence, and scientific computing in Python.

## Import Library

```python
import numpy as np
```

## Common Functions

1. **np.array()** – Creates a NumPy array.
2. **np.zeros()** – Creates an array filled with zeros.
3. **np.ones()** – Creates an array filled with ones.
4. **np.empty()** – Creates an empty array without initializing values.
5. **np.arange()** – Generates evenly spaced values within a range.
6. **np.linspace()** – Generates evenly spaced values between two limits.
7. **np.reshape()** – Changes the shape of an array.
8. **np.flatten()** – Converts a multidimensional array into a one-dimensional array.
9. **np.transpose()** – Transposes rows and columns of an array.
10. **np.mean()** – Calculates the average value.
11. **np.median()** – Calculates the median value.
12. **np.std()** – Calculates the standard deviation.
13. **np.var()** – Calculates the variance.
14. **np.sum()** – Returns the sum of array elements.
15. **np.min()** – Returns the minimum value.
16. **np.max()** – Returns the maximum value.
17. **np.argmin()** – Returns the index of the minimum value.
18. **np.argmax()** – Returns the index of the maximum value.
19. **np.sort()** – Sorts the elements of an array.
20. **np.unique()** – Returns unique values from an array.
21. **np.concatenate()** – Joins two or more arrays.
22. **np.split()** – Splits an array into multiple sub-arrays.
23. **np.random.rand()** – Generates random decimal numbers.
24. **np.random.randint()** – Generates random integers within a specified range.
25. **np.dot()** – Performs matrix multiplication (dot product).

````python
import numpy as np

# 1. np.array()
arr = np.array([10, 20, 30, 40, 50])
print("1. Array:", arr)

# 2. np.zeros()
print("\n2. Zeros Array:\n", np.zeros((2,3)))

# 3. np.ones()
print("\n3. Ones Array:\n", np.ones((2,2)))

# 4. np.empty()
print("\n4. Empty Array:\n", np.empty((2,2)))

# 5. np.arange()
print("\n5. Arange:", np.arange(1,11,2))

# 6. np.linspace()
print("\n6. Linspace:", np.linspace(0,10,5))

# 7. np.reshape()
a = np.arange(1,13)
print("\n7. Reshape:\n", a.reshape(3,4))

# 8. np.flatten()
print("\n8. Flatten:", a.reshape(3,4).flatten())

# 9. np.transpose()
matrix = np.array([[1,2],[3,4]])
print("\n9. Transpose:\n", matrix.T)

# 10. np.mean()
print("\n10. Mean:", np.mean(arr))

# 11. np.median()
print("11. Median:", np.median(arr))

# 12. np.std()
print("12. Standard Deviation:", np.std(arr))

# 13. np.var()
print("13. Variance:", np.var(arr))

# 14. np.sum()
print("14. Sum:", np.sum(arr))

# 15. np.min()
print("15. Minimum:", np.min(arr))

# 16. np.max()
print("16. Maximum:", np.max(arr))

# 17. np.argmin()
print("17. Index of Minimum:", np.argmin(arr))

# 18. np.argmax()
print("18. Index of Maximum:", np.argmax(arr))

# 19. np.sort()
unsorted = np.array([5,2,8,1,9])
print("19. Sorted:", np.sort(unsorted))

# 20. np.unique()
dup = np.array([1,2,2,3,4,4,5])
print("20. Unique:", np.unique(dup))

# 21. np.concatenate()
a1 = np.array([1,2,3])
a2 = np.array([4,5,6])
print("21. Concatenate:", np.concatenate((a1,a2)))

# 22. np.split()
arr2 = np.arange(8)
print("22. Split:", np.split(arr2,4))

# 23. np.random.rand()
print("23. Random Decimal:\n", np.random.rand(2,2))

# 24. np.random.randint()
print("24. Random Integer:\n", np.random.randint(1,100,(2,3)))

# 25. np.dot()
x = np.array([[1,2],[3,4]])
y = np.array([[5,6],[7,8]])
print("25. Dot Product:\n", np.dot(x,y))
````

# 2. Pandas

## Introduction

Pandas is a powerful open-source Python library used for data manipulation and analysis. It provides two primary data structures: **Series** and **DataFrame**, which make it easy to organize and analyze structured data. Pandas allows users to read and write data from various file formats such as CSV, Excel, JSON, SQL databases, and more. It offers powerful functions for cleaning, filtering, sorting, merging, grouping, and transforming datasets. Pandas integrates seamlessly with NumPy, Matplotlib, Seaborn, and Scikit-Learn, making it one of the most important libraries in Data Science and Machine Learning. Due to its efficiency and ease of use, Pandas is widely used for data preprocessing before building Machine Learning models.

## Import Library

```python
import pandas as pd
```

## Common Functions

1. **pd.DataFrame()** – Creates a DataFrame.
2. **pd.Series()** – Creates a Series.
3. **head()** – Displays the first five rows.
4. **tail()** – Displays the last five rows.
5. **info()** – Displays dataset information.
6. **describe()** – Generates statistical summary.
7. **shape** – Returns the number of rows and columns.
8. **columns** – Displays column names.
9. **index** – Displays row indexes.
10. **dtypes** – Shows data types of each column.
11. **isnull()** – Detects missing values.
12. **fillna()** – Replaces missing values.
13. **dropna()** – Removes missing values.
14. **drop()** – Removes rows or columns.
15. **rename()** – Renames columns or indexes.
16. **sort_values()** – Sorts data by column values.
17. **sort_index()** – Sorts data by index.
18. **groupby()** – Groups data based on column values.
19. **value_counts()** – Counts unique values.
20. **unique()** – Returns unique values.
21. **duplicated()** – Detects duplicate rows.
22. **drop_duplicates()** – Removes duplicate rows.
23. **sample()** – Returns random rows.
24. **concat()** – Combines DataFrames.
25. **merge()** – Joins DataFrames using common columns.

````python
import pandas as pd

# Sample DataFrame
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Alice"],
    "Age": [20, 22, 21, 23, 20],
    "Marks": [85, 90, 88, 92, 85]
}

df = pd.DataFrame(data)

# 1. DataFrame()
print("1. DataFrame:\n", df)

# 2. Series()
print("\n2. Series:\n", pd.Series([10,20,30]))

# 3. head()
print("\n3. Head:\n", df.head())

# 4. tail()
print("\n4. Tail:\n", df.tail())

# 5. info()
print("\n5. Info:")
df.info()

# 6. describe()
print("\n6. Describe:\n", df.describe())

# 7. shape
print("\n7. Shape:", df.shape)

# 8. columns
print("\n8. Columns:", df.columns)

# 9. index
print("\n9. Index:", df.index)

# 10. dtypes
print("\n10. Data Types:\n", df.dtypes)

# 11. isnull()
print("\n11. Missing Values:\n", df.isnull())

# 12. fillna()
df2 = df.copy()
df2.loc[1, "Marks"] = None
print("\n12. Fill Missing Values:\n", df2.fillna(0))

# 13. dropna()
print("\n13. Drop Missing Values:\n", df2.dropna())

# 14. drop()
print("\n14. Drop Column:\n", df.drop(columns=["Marks"]))

# 15. rename()
print("\n15. Rename Column:\n", df.rename(columns={"Marks":"Score"}))

# 16. sort_values()
print("\n16. Sort by Age:\n", df.sort_values(by="Age"))

# 17. sort_index()
print("\n17. Sort Index:\n", df.sort_index(ascending=False))

# 18. groupby()
print("\n18. Group By Name:\n", df.groupby("Name")["Marks"].mean())

# 19. value_counts()
print("\n19. Value Counts:\n", df["Name"].value_counts())

# 20. unique()
print("\n20. Unique Names:", df["Name"].unique())

# 21. duplicated()
print("\n21. Duplicated Rows:\n", df.duplicated())

# 22. drop_duplicates()
print("\n22. Remove Duplicates:\n", df.drop_duplicates())

# 23. sample()
print("\n23. Random Sample:\n", df.sample(2))

# 24. concat()
df_new = pd.DataFrame({"Name":["Eva"],"Age":[24],"Marks":[95]})
print("\n24. Concatenate:\n", pd.concat([df, df_new], ignore_index=True))

# 25. merge()
student = pd.DataFrame({
    "ID":[1,2,3],
    "Name":["Alice","Bob","Charlie"]
})

marks = pd.DataFrame({
    "ID":[1,2,3],
    "Marks":[85,90,88]
})

print("\n25. Merge:\n", pd.merge(student, marks, on="ID"))
````

````python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1,2,3,4,5])
y = np.array([2,4,6,8,10])

# 1. plot()
plt.figure()
plt.plot(x, y)
plt.title("1. Line Plot")
plt.show()

# 2. scatter()
plt.figure()
plt.scatter(x, y)
plt.title("2. Scatter Plot")
plt.show()

# 3. bar()
plt.figure()
plt.bar(x, y)
plt.title("3. Bar Chart")
plt.show()

# 4. barh()
plt.figure()
plt.barh(x, y)
plt.title("4. Horizontal Bar Chart")
plt.show()

# 5. hist()
plt.figure()
plt.hist([1,2,2,3,3,3,4,4,5])
plt.title("5. Histogram")
plt.show()

# 6. pie()
plt.figure()
plt.pie([30,40,30], labels=["A","B","C"], autopct="%1.1f%%")
plt.title("6. Pie Chart")
plt.show()

# 7. boxplot()
plt.figure()
plt.boxplot([10,20,15,30,25,35])
plt.title("7. Box Plot")
plt.show()

# 8-16. Labels, title, legend, grid, limits and ticks
plt.figure()
plt.plot(x, y, label="Data")
plt.xlabel("X-axis")           # 8
plt.ylabel("Y-axis")           # 9
plt.title("8-16 Functions")    # 10
plt.legend()                   # 11
plt.grid(True)                 # 12
plt.xlim(0,6)                  # 13
plt.ylim(0,12)                 # 14
plt.xticks(x)                  # 15
plt.yticks(range(0,13,2))      # 16
plt.show()

# 17. figure()
plt.figure(figsize=(6,4))
plt.plot(x,y)
plt.title("17. Figure")
plt.show()

# 18. subplot()
plt.figure(figsize=(8,3))
plt.subplot(1,2,1)
plt.plot(x,y)
plt.title("Plot")
plt.subplot(1,2,2)
plt.bar(x,y)
plt.title("Bar")
plt.show()

# 19. savefig()
plt.figure()
plt.plot(x,y)
plt.title("19. Save Figure")
plt.savefig("graph.png")
plt.show()

# 20. show()
plt.figure()
plt.plot(x,y)
plt.title("20. Show Graph")
plt.show()

# 21. fill_between()
plt.figure()
plt.plot(x,y)
plt.fill_between(x,y,color="lightblue")
plt.title("21. Fill Between")
plt.show()

# 22. stem()
plt.figure()
plt.stem(x,y)
plt.title("22. Stem Plot")
plt.show()

# 23. step()
plt.figure()
plt.step(x,y)
plt.title("23. Step Plot")
plt.show()

# 24. errorbar()
plt.figure()
plt.errorbar(x,y,yerr=0.5,fmt='o')
plt.title("24. Error Bar")
plt.show()

# 25. text()
plt.figure()
plt.plot(x,y)
plt.text(3,6,"Center Point")
plt.title("25. Text")
plt.show()
````

# 5. Scikit-Learn (sklearn)

## Introduction

Scikit-Learn, commonly known as **sklearn**, is one of the most widely used Machine Learning libraries in Python. It provides efficient and easy-to-use tools for building Machine Learning models for classification, regression, clustering, and dimensionality reduction. Built on top of NumPy, SciPy, and Matplotlib, it offers a consistent interface for training, testing, and evaluating models. Scikit-Learn also includes utilities for data preprocessing, feature selection, model selection, and performance evaluation. Due to its simplicity, extensive documentation, and rich collection of algorithms, Scikit-Learn is widely used in education, research, and industry for Machine Learning applications.

## Import Library

```python
from sklearn.model_selection import train_test_split
```

## Common Functions

1. **train_test_split()** – Splits data into training and testing sets.
2. **fit()** – Trains the Machine Learning model.
3. **predict()** – Predicts output for new data.
4. **score()** – Evaluates the model performance.
5. **transform()** – Transforms input data.
6. **fit_transform()** – Fits and transforms data together.
7. **inverse_transform()** – Converts transformed data back to original.
8. **accuracy_score()** – Calculates classification accuracy.
9. **confusion_matrix()** – Generates a confusion matrix.
10. **classification_report()** – Displays classification metrics.
11. **mean_squared_error()** – Calculates Mean Squared Error.
12. **mean_absolute_error()** – Calculates Mean Absolute Error.
13. **r2_score()** – Calculates R² score.
14. **StandardScaler()** – Standardizes features.
15. **MinMaxScaler()** – Scales features between 0 and 1.
16. **LabelEncoder()** – Encodes categorical labels.
17. **OneHotEncoder()** – Converts categorical variables into one-hot vectors.
18. **LinearRegression()** – Creates a Linear Regression model.
19. **LogisticRegression()** – Creates a Logistic Regression model.
20. **DecisionTreeClassifier()** – Creates a Decision Tree classifier.
21. **RandomForestClassifier()** – Creates a Random Forest classifier.
22. **KNeighborsClassifier()** – Creates a KNN classifier.
23. **SVC()** – Creates a Support Vector Machine classifier.
24. **KMeans()** – Performs K-Means clustering.
25. **PCA()** – Performs Principal Component Analysis.


````python
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler, LabelEncoder, OneHotEncoder
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    mean_squared_error,
    mean_absolute_error,
    r2_score
)

# Sample Data
X = np.array([[1],[2],[3],[4],[5],[6],[7],[8]])
y = np.array([2,4,6,8,10,12,14,16])

# 1. train_test_split()
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)
print("1. Train/Test Split Done")

# 2. fit()
model = LinearRegression()
model.fit(X_train, y_train)
print("2. Model Trained")

# 3. predict()
pred = model.predict(X_test)
print("3. Prediction:", pred)

# 4. score()
print("4. Score:", model.score(X_test, y_test))

# 5. transform()
scaler = StandardScaler()
scaler.fit(X_train)
print("5. Transform:\n", scaler.transform(X_train))

# 6. fit_transform()
scaled = scaler.fit_transform(X_train)
print("6. Fit Transform:\n", scaled)

# 7. inverse_transform()
print("7. Inverse Transform:\n", scaler.inverse_transform(scaled))

# 8. accuracy_score()
y_true = [0,1,1,0]
y_pred = [0,1,0,0]
print("8. Accuracy:", accuracy_score(y_true,y_pred))

# 9. confusion_matrix()
print("9.\n", confusion_matrix(y_true,y_pred))

# 10. classification_report()
print("10.\n", classification_report(y_true,y_pred))

# 11. mean_squared_error()
print("11. MSE:", mean_squared_error(y_test,pred))

# 12. mean_absolute_error()
print("12. MAE:", mean_absolute_error(y_test,pred))

# 13. r2_score()
print("13. R2 Score:", r2_score(y_test,pred))

# 14. StandardScaler()
std = StandardScaler()
print("14.\n", std.fit_transform(X))

# 15. MinMaxScaler()
mm = MinMaxScaler()
print("15.\n", mm.fit_transform(X))

# 16. LabelEncoder()
le = LabelEncoder()
print("16.", le.fit_transform(["Cat","Dog","Cat","Bird"]))

# 17. OneHotEncoder()
ohe = OneHotEncoder(sparse_output=False)
print("17.\n", ohe.fit_transform([["Red"],["Blue"],["Green"]]))

# 18. LinearRegression()
lr = LinearRegression()
lr.fit(X,y)
print("18.", lr.predict([[9]]))

# 19. LogisticRegression()
X_cls = [[1],[2],[3],[4]]
y_cls = [0,0,1,1]
log = LogisticRegression()
log.fit(X_cls,y_cls)
print("19.", log.predict([[2.5]]))

# 20. DecisionTreeClassifier()
dt = DecisionTreeClassifier(random_state=0)
dt.fit(X_cls,y_cls)
print("20.", dt.predict([[3]]))

# 21. RandomForestClassifier()
rf = RandomForestClassifier(random_state=0)
rf.fit(X_cls,y_cls)
print("21.", rf.predict([[3]]))

# 22. KNeighborsClassifier()
knn = KNeighborsClassifier(n_neighbors=1)
knn.fit(X_cls,y_cls)
print("22.", knn.predict([[3]]))

# 23. SVC()
svc = SVC()
svc.fit(X_cls,y_cls)
print("23.", svc.predict([[3]]))

# 24. KMeans()
kmeans = KMeans(n_clusters=2, random_state=0, n_init=10)
kmeans.fit(X)
print("24. Cluster Labels:", kmeans.labels_)

# 25. PCA()
X_pca = np.array([[1,2],[2,3],[3,4],[4,5]])
pca = PCA(n_components=1)
print("25.\n", pca.fit_transform(X_pca))
````

# 7. Keras

## Introduction

Keras is a high-level Deep Learning API integrated with TensorFlow. It provides a simple, user-friendly interface for building, training, and evaluating neural network models. ## Import Library

```python
from tensorflow import keras
```

## Common Functions

1. **Sequential()** – Creates a sequential neural network model.
2. **Dense()** – Adds a fully connected (dense) layer.
3. **Input()** – Defines the input layer.
4. **Dropout()** – Prevents overfitting by randomly dropping neurons.
5. **Flatten()** – Converts multidimensional input into a single dimension.
6. **Activation()** – Applies an activation function.
7. **compile()** – Configures the model for training.
8. **fit()** – Trains the model.
9. **evaluate()** – Evaluates the model performance.
10. **predict()** – Predicts output for new data.
11. **summary()** – Displays the model architecture.
12. **save()** – Saves the trained model.
13. **load_model()** – Loads a saved model.
14. **get_weights()** – Returns model weights.
15. **set_weights()** – Sets model weights.
16. **add()** – Adds a layer to the model.
17. **pop()** – Removes the last layer.
18. **BatchNormalization()** – Normalizes layer outputs.
19. **Conv2D()** – Adds a 2D convolution layer.
20. **MaxPooling2D()** – Adds a max pooling layer.
21. **SimpleRNN()** – Adds a Simple RNN layer.

23. **Embedding()** – Converts integer inputs into dense vectors.
24. **EarlyStopping()** – Stops training when performance stops improving.
25. **ModelCheckpoint()** – Saves the best model during training.

````python
from tensorflow import keras
from tensorflow.keras.models import load_model
from tensorflow.keras.layers import (
    Dense, Input, Dropout, Flatten,
    Activation, BatchNormalization,
    Conv2D, MaxPooling2D,
    SimpleRNN, LSTM, Embedding
)
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
import numpy as np

# 1. Sequential()
model = keras.Sequential()
print("1. Sequential Model Created")

# 2. Dense()
model.add(Dense(8, input_shape=(4,)))
print("2. Dense Layer Added")

# 3. Input()
model2 = keras.Sequential([
    Input(shape=(4,)),
    Dense(8)
])
print("3. Input Layer Created")

# 4. Dropout()
model.add(Dropout(0.2))
print("4. Dropout Layer Added")

# 5. Flatten()
model3 = keras.Sequential([
    Flatten(input_shape=(2,2))
])
print("5. Flatten Layer Added")

# 6. Activation()
model.add(Activation("relu"))
print("6. Activation Layer Added")

# 7. compile()
model.add(Dense(1))
model.compile(optimizer="adam",
              loss="mse",
              metrics=["mae"])
print("7. Model Compiled")

# Sample Data
X = np.random.rand(20,4)
Y = np.random.rand(20)

# 8. fit()
model.fit(X,Y,epochs=1,verbose=0)
print("8. Model Trained")

# 9. evaluate()
print("9. Evaluation:", model.evaluate(X,Y,verbose=0))

# 10. predict()
print("10. Prediction:\n", model.predict(X[:2],verbose=0))

# 11. summary()
print("11. Model Summary")
model.summary()

# 12. save()
model.save("keras_model.keras")
print("12. Model Saved")

# 13. load_model()
loaded_model = load_model("keras_model.keras")
print("13. Model Loaded")

# 14. get_weights()
weights = model.get_weights()
print("14. Number of Weight Arrays:", len(weights))

# 15. set_weights()
loaded_model.set_weights(weights)
print("15. Weights Set")

# 16. add()
model.add(Dense(2))
print("16. New Layer Added")

# 17. pop()
model.pop()
print("17. Last Layer Removed")

# 18. BatchNormalization()
bn_model = keras.Sequential([
    Dense(8,input_shape=(4,)),
    BatchNormalization()
])
print("18. BatchNormalization Added")

# 19. Conv2D()
cnn = keras.Sequential([
    Conv2D(8,(3,3),input_shape=(28,28,1))
])
print("19. Conv2D Layer Added")

# 20. MaxPooling2D()
cnn.add(MaxPooling2D((2,2)))
print("20. MaxPooling2D Added")

# 21. SimpleRNN()
rnn = keras.Sequential([
    SimpleRNN(8,input_shape=(5,3))
])
print("21. SimpleRNN Added")

# 22. LSTM()
lstm = keras.Sequential([
    LSTM(8,input_shape=(5,3))
])
print("22. LSTM Added")

# 23. Embedding()
embed = keras.Sequential([
    Embedding(input_dim=100,
              output_dim=8,
              input_length=10)
])
print("23. Embedding Layer Added")

# 24. EarlyStopping()
early = EarlyStopping(
    monitor="loss",
    patience=2
)
print("24. EarlyStopping Callback Created")

# 25. ModelCheckpoint()
checkpoint = ModelCheckpoint(
    "best_model.keras",
    save_best_only=True
)
print("25. ModelCheckpoint Created")
````

```python
from scipy import stats
```

## Common Functions

1. **stats.describe()** – Returns descriptive statistics of data.
2. **stats.tmean()** – Calculates the trimmed mean.
3. **stats.mode()** – Finds the most frequent value.
4. **stats.median_abs_deviation()** – Calculates the median absolute deviation.
5. **stats.norm.pdf()** – Computes the probability density function.
6. **stats.norm.cdf()** – Computes the cumulative distribution function.
7. **stats.norm.rvs()** – Generates random samples from a normal distribution.
8. **stats.zscore()** – Computes Z-scores.
9. **stats.percentileofscore()** – Calculates the percentile rank.
10. **stats.rankdata()** – Assigns ranks to data values.
11. **linalg.det()** – Calculates the determinant of a matrix.
12. **linalg.inv()** – Computes the inverse of a matrix.
13. **linalg.eig()** – Computes eigenvalues and eigenvectors.
14. **linalg.solve()** – Solves linear equations.
15. **integrate.quad()** – Performs numerical integration.
16. **optimize.minimize()** – Finds the minimum of a function.
17. **fft.fft()** – Computes the Fast Fourier Transform.
18. **fft.ifft()** – Computes the Inverse Fourier Transform.
19. **interpolate.interp1d()** – Performs one-dimensional interpolation.
20. **signal.find_peaks()** – Finds peaks in a signal.
21. **signal.convolve()** – Performs convolution of signals.
22. **io.savemat()** – Saves data in MATLAB (.mat) format.
23. **io.loadmat()** – Loads MATLAB (.mat) files.
24. **special.factorial()** – Computes the factorial of numbers.
25. **special.comb()** – Computes combinations (nCr).

````python
from scipy import stats, linalg, integrate, optimize, fft, interpolate, signal, io, special
import numpy as np

data = [2, 4, 4, 5, 6, 8, 9]

# 1. stats.describe()
print("1.", stats.describe(data))

# 2. stats.tmean()
print("2.", stats.tmean(data))

# 3. stats.mode()
print("3.", stats.mode(data))

# 4. stats.median_abs_deviation()
print("4.", stats.median_abs_deviation(data))

# 5. stats.norm.pdf()
print("5.", stats.norm.pdf(0))

# 6. stats.norm.cdf()
print("6.", stats.norm.cdf(0))

# 7. stats.norm.rvs()
print("7.", stats.norm.rvs(size=5))

# 8. stats.zscore()
print("8.", stats.zscore(data))

# 9. stats.percentileofscore()
print("9.", stats.percentileofscore(data, 5))

# 10. stats.rankdata()
print("10.", stats.rankdata(data))

# Matrix for linear algebra
A = np.array([[2,1],[5,3]])

# 11. linalg.det()
print("11.", linalg.det(A))

# 12. linalg.inv()
print("12.\n", linalg.inv(A))

# 13. linalg.eig()
values, vectors = linalg.eig(A)
print("13. Eigenvalues:", values)

# 14. linalg.solve()
b = np.array([3,7])
print("14.", linalg.solve(A,b))

# 15. integrate.quad()
result, error = integrate.quad(lambda x: x**2, 0, 2)
print("15.", result)

# 16. optimize.minimize()
res = optimize.minimize(lambda x: x**2 + 4*x + 4, x0=0)
print("16.", res.x)

# 17. fft.fft()
arr = np.array([1,2,3,4])
print("17.", fft.fft(arr))

# 18. fft.ifft()
print("18.", fft.ifft(fft.fft(arr)))

# 19. interpolate.interp1d()
x = np.array([0,1,2,3])
y = np.array([0,2,4,6])
f = interpolate.interp1d(x,y)
print("19.", f(1.5))

# 20. signal.find_peaks()
peaks, _ = signal.find_peaks([1,3,2,5,1,4])
print("20.", peaks)

# 21. signal.convolve()
print("21.", signal.convolve([1,2,3],[1,1]))

# 22. io.savemat()
io.savemat("sample.mat", {"array": arr})
print("22. sample.mat file saved")

# 23. io.loadmat()
loaded = io.loadmat("sample.mat")
print("23.", loaded["array"])

# 24. special.factorial()
print("24.", special.factorial(5))

# 25. special.comb()
print("25.", special.comb(5,2))
````
