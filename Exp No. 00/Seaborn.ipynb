# 4. Seaborn

## Introduction

Seaborn is a high-level Python data visualization library built on top of Matplotlib. It provides an easy and attractive way to create statistical graphics with minimal code. Seaborn integrates well with Pandas DataFrames and includes built-in datasets for learning and analysis. It is widely used for Exploratory Data Analysis (EDA), helping users understand relationships, distributions, trends, and correlations within data. The library offers beautiful default themes and advanced visualization functions such as heatmaps, pair plots, violin plots, and regression plots. Due to its simplicity and powerful visualization capabilities, Seaborn has become one of the most popular libraries in Data Science and Machine Learning.

## Import Library

```python
import seaborn as sns
```

## Common Functions

1. **load_dataset()** – Loads built-in sample datasets.
2. **scatterplot()** – Creates a scatter plot.
3. **lineplot()** – Creates a line graph.
4. **barplot()** – Creates a bar chart.
5. **countplot()** – Displays the count of categorical data.
6. **histplot()** – Creates a histogram.
7. **boxplot()** – Creates a box plot.
8. **violinplot()** – Creates a violin plot.
9. **stripplot()** – Creates a strip plot.
10. **swarmplot()** – Creates a swarm plot.
11. **pairplot()** – Displays pairwise relationships.
12. **heatmap()** – Creates a heatmap.
13. **jointplot()** – Shows joint distribution of two variables.
14. **kdeplot()** – Creates a Kernel Density Estimate plot.
15. **displot()** – Displays distribution plots.
16. **relplot()** – Creates relational plots.
17. **catplot()** – Creates categorical plots.
18. **lmplot()** – Creates a regression plot.
19. **regplot()** – Displays a regression line.
20. **ecdfplot()** – Creates an ECDF plot.
21. **rugplot()** – Displays rug marks.
22. **set_theme()** – Sets the default theme.
23. **set_style()** – Changes plot style.
24. **set_palette()** – Changes color palette.
25. **color_palette()** – Displays or creates a color palette.

````python
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
iris = sns.load_dataset("iris")

# 1. load_dataset()
print("1. Dataset Loaded")
print(iris.head())

# 2. scatterplot()
sns.scatterplot(data=iris, x="sepal_length", y="sepal_width")
plt.title("2. Scatter Plot")
plt.show()

# 3. lineplot()
sns.lineplot(data=iris, x="sepal_length", y="petal_length")
plt.title("3. Line Plot")
plt.show()

# 4. barplot()
sns.barplot(data=iris, x="species", y="petal_length")
plt.title("4. Bar Plot")
plt.show()

# 5. countplot()
sns.countplot(data=iris, x="species")
plt.title("5. Count Plot")
plt.show()

# 6. histplot()
sns.histplot(data=iris, x="petal_length")
plt.title("6. Histogram")
plt.show()

# 7. boxplot()
sns.boxplot(data=iris, x="species", y="petal_length")
plt.title("7. Box Plot")
plt.show()

# 8. violinplot()
sns.violinplot(data=iris, x="species", y="petal_length")
plt.title("8. Violin Plot")
plt.show()

# 9. stripplot()
sns.stripplot(data=iris, x="species", y="petal_length")
plt.title("9. Strip Plot")
plt.show()

# 10. swarmplot()
sns.swarmplot(data=iris, x="species", y="petal_length")
plt.title("10. Swarm Plot")
plt.show()

# 11. pairplot()
sns.pairplot(iris)
plt.show()

# 12. heatmap()
sns.heatmap(iris.select_dtypes("number").corr(), annot=True)
plt.title("12. Heatmap")
plt.show()

# 13. jointplot()
sns.jointplot(data=iris, x="sepal_length", y="petal_length")
plt.show()

# 14. kdeplot()
sns.kdeplot(data=iris, x="petal_length", fill=True)
plt.title("14. KDE Plot")
plt.show()

# 15. displot()
sns.displot(data=iris, x="petal_width")
plt.show()

# 16. relplot()
sns.relplot(data=iris, x="sepal_length", y="petal_width", hue="species")
plt.show()

# 17. catplot()
sns.catplot(data=iris, x="species", y="petal_length", kind="box")
plt.show()

# 18. lmplot()
sns.lmplot(data=iris, x="sepal_length", y="petal_length")
plt.show()

# 19. regplot()
sns.regplot(data=iris, x="sepal_length", y="petal_length")
plt.title("19. Regression Plot")
plt.show()

# 20. ecdfplot()
sns.ecdfplot(data=iris, x="petal_length")
plt.title("20. ECDF Plot")
plt.show()

# 21. rugplot()
sns.rugplot(data=iris, x="petal_length")
plt.title("21. Rug Plot")
plt.show()

# 22. set_theme()
sns.set_theme()

# 23. set_style()
sns.set_style("darkgrid")
sns.scatterplot(data=iris, x="sepal_length", y="sepal_width")
plt.title("23. Darkgrid Style")
plt.show()

# 24. set_palette()
sns.set_palette("Set2")
sns.barplot(data=iris, x="species", y="petal_length")
plt.title("24. Set Palette")
plt.show()

# 25. color_palette()
print("25. Current Color Palette:")
print(sns.color_palette())
````
